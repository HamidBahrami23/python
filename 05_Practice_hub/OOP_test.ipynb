```python
class Animal():
    zoo_name = "Africa"
    def __init__(self , name , species , age , sound):
        self.name = name
        self.species = species
        self.age = age
        self.sound = sound
        print("animal created")
    def info(self):
        print(f"the animal {self.name} from {self.species} species, with {self.age} age and {self.sound} sound created")
    def make_sound(self):
        print(f"The animal sound is: {self.sound}")
```

```python
a1 = Animal("tiger" , 2 , 12 , "HP")
```

```python
a1.info()
```

```python
a1.make_sound()
```
